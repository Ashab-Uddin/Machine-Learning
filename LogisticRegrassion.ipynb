```python
# Logistic Regression using Gradient Descent from Scratch

import numpy as np
import matplotlib.pyplot as plt

# Small dataset
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([0, 0, 0, 1, 1], dtype=float)

# Initialize weight and bias
w = 0
b = 0

# Learning rate and number of iterations
learning_rate = 0.1
epochs = 10000

# Number of samples
m = len(X)

# Store loss values
loss_history = []

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Gradient Descent loop
for i in range(epochs):

    # Linear score
    z = w * X + b

    # Predicted probability
    y_pred = sigmoid(z)

    # Binary cross-entropy loss
    loss = -(1 / m) * np.sum(
        y * np.log(y_pred) + (1 - y) * np.log(1 - y_pred)
    )
    loss_history.append(loss)

    # Calculate gradients
    dw = (1 / m) * np.sum((y_pred - y) * X)
    db = (1 / m) * np.sum(y_pred - y)

    # Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

# Display final parameters
print("Final weight:", w)
print("Final bias:", b)
print("Final loss:", loss_history[-1])

# Predicted probabilities
probabilities = sigmoid(w * X + b)

# Convert probabilities into class labels
predicted_classes = (probabilities >= 0.5).astype(int)

print("\nActual classes:")
print(y.astype(int))

print("\nPredicted probabilities:")
print(probabilities)

print("\nPredicted classes:")
print(predicted_classes)

# Plot loss curve
plt.figure(figsize=(7, 5))
plt.plot(range(epochs), loss_history)
plt.xlabel("Iterations")
plt.ylabel("Log Loss")
plt.title("Loss Reduction During Gradient Descent")
plt.grid(True)
plt.show()

# Plot sigmoid curve with data points
x_values = np.linspace(0, 6, 100)
y_values = sigmoid(w * x_values + b)

plt.figure(figsize=(7, 5))
plt.scatter(X, y, color="red", label="Actual Data")
plt.plot(x_values, y_values, color="blue", label="Sigmoid Curve")
plt.axhline(y=0.5, color="green", linestyle="--", label="Decision Threshold")

plt.xlabel("Study Hours")
plt.ylabel("Probability of Passing")
plt.title("Logistic Regression using Gradient Descent")
plt.legend()
plt.grid(True)
plt.show()
```
